### 10. Membuat Aplikasi Dashboard Streamlit (`app.py`)

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

st.set_page_config(layout="wide", page_title="Dashboard Analisis Data Global")
st.title("📊 Dashboard Analisis Data & Visualisasi Interaktif")
st.markdown("Dashboard ini menggabungkan seluruh visualisasi interaktif yang telah dibuat sebelumnya.")

# --- RE-CREATE DATASETS ---
# 1. Bar Chart Data
df_bar = pd.DataFrame({
    "Kuartal": ["Q1", "Q2", "Q3", "Q4"] * 3,
    "Produk": ["Cloud Server"]*4 + ["CyberSec Suite"]*4 + ["AI Analytics"]*4,
    "Penjualan": [450, 520, 490, 610, 320, 410, 460, 530, 280, 350, 440, 590]
})

# 2. Pie Chart Data
df_pie = pd.DataFrame({
    "Perusahaan": ["Nusantara Cloud", "Garuda Tech", "Sunda Data", "Borneo Host", "Lainnya"],
    "Pangsa_Pasar": [34.5, 26.0, 18.5, 12.0, 9.0]
})

# 3. Line Chart Data
df_line = pd.DataFrame({
    "Tahun": [2020, 2021, 2022, 2023, 2024, 2025, 2026],
    "Unit_Terjual": [12500, 15800, 14900, 19200, 24500, 28900, 34200]
})

# 4. Area Chart Data
df_area = pd.DataFrame({
    "Periode": ["2024-Q1", "2024-Q2", "2024-Q3", "2024-Q4", "2025-Q1", "2025-Q2", "2025-Q3", "2025-Q4"],
    "Pendapatan": [12.4, 14.1, 13.8, 16.5, 18.2, 20.4, 22.1, 25.8]
})

# 5. Scatter Plot Data
np.random.seed(42)
harga = np.random.uniform(50, 500, 80)
penjualan = 1200 - (1.8 * harga) + np.random.normal(0, 85, 80)
df_scatter = pd.DataFrame({"Harga_Ribu_Rp": harga, "Volume_Penjualan": np.maximum(penjualan, 20)})

# 6. Histogram Data
np.random.seed(7)
gaji_karyawan = np.random.lognormal(mean=2.2, sigma=0.45, size=500)

# 7. Box Plot Data
np.random.seed(10)
df_box = pd.DataFrame({
  "Kelas": np.repeat(["IF-7A", "IF-7B", "IF-7C", "IF-7D"], 40),
  "Nilai_Ujian": np.concatenate([
    np.random.normal(78, 8, 40),
    np.random.normal(72, 12, 40),
    np.concatenate([np.random.normal(82, 6, 38), [35, 40]]),
    np.random.normal(75, 10, 40)
    ])
})
df_box["Nilai_Ujian"] = df_box["Nilai_Ujian"].clip(0, 100)

# 8. Heatmap Data
data_suhu = pd.DataFrame({
  "Jan": [26.5, 1.2, 3.5, 22.8, 28.1],
  "Apr": [27.2, 11.4, 14.8, 19.5, 29.4],
  "Jul": [26.8, 22.6, 26.2, 13.2, 28.9],
  "Okt": [27.1, 12.8, 17.5, 18.4, 28.5]
}, index=["Jakarta", "London", "Tokyo", "Sydney", "Singapura"])

# 9. Choropleth Data
df_geo = pd.DataFrame({
  "ISO_Alpha": ["IDN", "USA", "JPN", "AUS", "NOR", "BRA", "IND", "RUS", "EGY", "CAN"],
  "Negara": ["Indonesia", "Amerika Serikat", "Jepang", "Australia", "Norwegia", "Brasil", "India", "Rusia", "Mesir", "Kanada"],
  "Suhu_C": [27.2, 12.5, 15.1, 21.8, 2.4, 25.5, 25.0, -5.1, 23.1, -4.2]
})

# --- SIDEBAR FILTERS (DYNAMIC) ---
st.sidebar.header("⚙️ Filter Dinamis")

# Filter 1: Produk (Untuk Bar Chart)
selected_products = st.sidebar.multiselect(
    "Pilih Kategori Produk (Bar Chart):",
    options=df_bar["Produk"].unique(),
    default=df_bar["Produk"].unique()
)

# Filter 2: Range Harga (Untuk Scatter Plot)
min_price, max_price = st.sidebar.slider(
    "Range Harga Produk (Scatter Plot):",
    float(df_scatter["Harga_Ribu_Rp"].min()),
    float(df_scatter["Harga_Ribu_Rp"].max()),
    (float(df_scatter["Harga_Ribu_Rp"].min()), float(df_scatter["Harga_Ribu_Rp"].max()))
)

# Filter 3: Kelas (Untuk Box Plot)
selected_classes = st.sidebar.multiselect(
    "Pilih Kelas (Box Plot):",
    options=df_box["Kelas"].unique(),
    default=df_box["Kelas"].unique()
)

# --- TABS LAYOUT ---
tab1, tab2, tab3, tab4, tab5, tab6, tab7 = st.tabs([
    "📈 Penjualan & Pasar",
    "💰 Tren Pendapatan",
    "🔍 Analisis Hubungan",
    "👥 Distribusi Gaji",
    "🎓 Nilai Mahasiswa",
    "🌡️ Suhu Global (Heatmap)",
    "🗺️ Peta Geografis"
])

with tab1:
    st.header("Analisis Penjualan & Pangsa Pasar")
    col1, col2 = st.columns(2)

    with col1:
        st.subheader("Perbandingan Penjualan Produk per Kuartal")
        filtered_df_bar = df_bar[df_bar["Produk"].isin(selected_products)]
        fig_bar = px.bar(
            filtered_df_bar,
            x="Kuartal",
            y="Penjualan",
            color="Produk",
            barmode="group",
            color_discrete_sequence=px.colors.qualitative.Prism,
            labels={"Penjualan": "Total Penjualan (Juta Rp)", "Kuartal": "Periode Kuartal"}
        )
        st.plotly_chart(fig_bar, use_container_width=True)

    with col2:
        st.subheader("Proporsi Pangsa Pasar Perusahaan")
        fig_pie = px.pie(
            df_pie,
            values="Pangsa_Pasar",
            names="Perusahaan",
            color_discrete_sequence=px.colors.qualitative.Pastel
        )
        fig_pie.update_traces(textposition='inside', textinfo='percent+label')
        st.plotly_chart(fig_pie, use_container_width=True)

with tab2:
    st.header("Tren Perkembangan Finansial")
    col1, col2 = st.columns(2)

    with col1:
        st.subheader("Tren Penjualan Produk (2020 - 2026)")
        fig_line = px.line(
            df_line,
            x="Tahun",
            y="Unit_Terjual",
            markers=True,
            labels={"Unit_Terjual": "Jumlah Unit Terjual"}
        )
        st.plotly_chart(fig_line, use_container_width=True)

    with col2:
        st.subheader("Tren Pendapatan Perusahaan per Kuartal")
        fig_area = px.area(
            df_area,
            x="Periode",
            y="Pendapatan",
            labels={"Pendapatan": "Pendapatan (Miliar Rp)", "Periode": "Kuartal"}
        )
        fig_area.update_traces(line_color="#2ca02c")
        st.plotly_chart(fig_area, use_container_width=True)

with tab3:
    st.header("Hubungan Antara Variabel")
    st.subheader("Korelasi Harga Produk terhadap Volume Penjualan")

    filtered_scatter = df_scatter[
        (df_scatter["Harga_Ribu_Rp"] >= min_price) &
        (df_scatter["Harga_Ribu_Rp"] <= max_price)
    ]

    fig_scatter = px.scatter(
        filtered_scatter,
        x="Harga_Ribu_Rp",
        y="Volume_Penjualan",
        trendline="ols",
        labels={"Harga_Ribu_Rp": "Harga Produk (Ribu Rp)", "Volume_Penjualan": "Volume Penjualan (Unit)"}
    )
    st.plotly_chart(fig_scatter, use_container_width=True)

with tab4:
    st.header("Analisis Distribusi Pendapatan")
    st.subheader("Distribusi Pendapatan Karyawan (n = 500)")
    fig_hist = px.histogram(
        x=gaji_karyawan,
        nbins=25,
        labels={"x": "Pendapatan Bulanan (Juta Rp)", "y": "Frekuensi"},
        color_discrete_sequence=["#9467bd"]
    )
    median_val = float(np.median(gaji_karyawan))
    fig_hist.add_vline(
        x=median_val,
        line_dash="dash",
        line_color="red",
        annotation_text=f"Median: {median_val:.1f} Juta"
    )
    st.plotly_chart(fig_hist, use_container_width=True)

with tab5:
    st.header("Distribusi Nilai Akademik")
    st.subheader("Distribusi Nilai Ujian Mahasiswa per Kelas")
    filtered_df_box = df_box[df_box["Kelas"].isin(selected_classes)]
    if not filtered_df_box.empty:
        fig_box = px.box(
            filtered_df_box,
            x="Kelas",
            y="Nilai_Ujian",
            color="Kelas",
            color_discrete_sequence=px.colors.qualitative.Set2,
            labels={"Nilai_Ujian": "Skor Ujian (0 - 100)"}
        )
        st.plotly_chart(fig_box, use_container_width=True)
    else:
        st.warning("Mohon pilih setidaknya satu kelas pada filter sidebar.")

with tab6:
    st.header("Matriks Suhu Global")
    st.subheader("Heatmap Rata-Rata Suhu Udara Kota Global (°C)")
    fig_heatmap = px.imshow(
        data_suhu,
        text_auto=".1f",
        color_continuous_scale="RdBu_r",
        labels={"x": "Bulan", "y": "Kota", "color": "Suhu (°C)"}
    )
    st.plotly_chart(fig_heatmap, use_container_width=True)

with tab7:
    st.header("Geospasial Suhu")
    st.subheader("Peta Distribusi Suhu Udara Global (°C)")
    fig_geo = px.choropleth(
        df_geo,
        locations="ISO_Alpha",
        color="Suhu_C",
        hover_name="Negara",
        color_continuous_scale=px.colors.sequential.YlOrRd,
        labels={"Suhu_C": "Suhu (°C)", "ISO_Alpha": "Kode Negara"}
    )
    st.plotly_chart(fig_geo, use_container_width=True)
```
